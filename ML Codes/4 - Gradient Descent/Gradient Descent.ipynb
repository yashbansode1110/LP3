```python
# Gradient Descent Algorithm

# Define learning rate
learning_rate = 0.1

# Starting point
x = 2

# Number of iterations
iterations = 50

# Gradient Descent
for i in range(iterations):

    # Calculate derivative
    gradient = 2 * (x + 3)

    # Update x
    x = x - learning_rate * gradient

    # Display result
    print("Iteration:", i + 1, "x =", x)

# Calculate minimum value
y = (x + 3) ** 2

print("\nMinimum x =", x)
print("Minimum y =", y)
```
